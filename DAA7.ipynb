```python
import matplotlib.pyplot as plt


# ==========================================
# CHECK WHETHER POSITION IS SAFE
# ==========================================

def is_safe(board, row, col):

    for prev_row in range(row):

        placed = board[prev_row]

        # Same column
        if placed == col:
            return False

        # Same diagonal
        if abs(prev_row - row) == abs(placed - col):
            return False

    return True


# ==========================================
# SOLVE N-QUEENS USING BACKTRACKING
# ==========================================

def solve_n_queens(n):

    board = [-1] * n

    solutions = []

    backtrack_count = [0]

    def backtrack(row):

        # All queens placed
        if row == n:
            solutions.append(board[:])
            return

        # Try every column
        for col in range(n):

            if is_safe(board, row, col):

                board[row] = col

                backtrack(row + 1)

                # Undo the move
                board[row] = -1

                backtrack_count[0] += 1

    backtrack(0)

    return solutions, backtrack_count[0]


# ==========================================
# DISPLAY CHESS BOARD
# ==========================================

def display_board(solution, n):

    print(" +" + "---+" * n)

    for row in range(n):

        print(" |", end="")

        for col in range(n):

            if solution[row] == col:
                print(" Q |", end="")
            else:
                print(" . |", end="")

        print()

        print(" +" + "---+" * n)


# ==========================================
# MAIN PROGRAM
# ==========================================

print("=" * 65)
print("       N-QUEENS PROBLEM USING BACKTRACKING")
print("=" * 65)


# Values of N
n_values = [4, 6, 8]

solution_counts = []
backtrack_counts = []


# Table heading
print("\n")
print(f"{'N':>8} {'Solutions':>15} {'Backtracks':>18}")
print("-" * 45)


for n in n_values:

    solutions, backtracks = solve_n_queens(n)

    solution_counts.append(len(solutions))
    backtrack_counts.append(backtracks)

    print(
        f"{n:>8}"
        f"{len(solutions):>15}"
        f"{backtracks:>18}"
    )


# ==========================================
# DISPLAY ALL 4-QUEENS SOLUTIONS
# ==========================================

solutions_4, backtracks_4 = solve_n_queens(4)

print("\n")
print("=" * 65)
print("             ALL 4-QUEENS SOLUTIONS")
print("=" * 65)


for i, solution in enumerate(solutions_4, 1):

    print(f"\nSolution {i}: {solution}")

    display_board(
        solution,
        4
    )


# ==========================================
# GRAPH 1 - NUMBER OF SOLUTIONS
# ==========================================

plt.figure(figsize=(9, 6))

plt.plot(
    n_values,
    solution_counts,
    marker="o",
    linewidth=2
)

plt.xlabel("N (Number of Queens)")
plt.ylabel("Number of Solutions")

plt.title(
    "N-Queens: Number of Solutions"
)

plt.xticks(n_values)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.show()


# ==========================================
# GRAPH 2 - NUMBER OF BACKTRACKS
# ==========================================

plt.figure(figsize=(9, 6))

plt.plot(
    n_values,
    backtrack_counts,
    marker="s",
    linewidth=2
)

plt.xlabel("N (Number of Queens)")
plt.ylabel("Number of Backtracks")

plt.title(
    "N-Queens: Backtracking Analysis"
)

plt.xticks(n_values)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.show()
```
